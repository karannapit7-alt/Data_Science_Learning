```python
# Filtering with Conditions.

df[df["Age"]>18]  # Simple Conditions.
```

```python
df[df["Age"]>21] ["Age"]  # Filtering only Age.
```

```python
# Filtering Student ID.

df[df["Student_ID"]>520]
```

```python
# Multiple Conditions (AND/OR)

df[(df['Student_ID']>520) & (df['Blood_Pressure_Systolic']>123)]   # This is AND Conditions.
```

```python
df[(df['Student_ID']>520) | (df['Blood_Pressure_Systolic']>123)]   # This is OR Conditions.
```

```python
# This is a cleaner, SQL- like way to filter.

df.query("Student_ID > 230")
```

```python
df.query("Student_ID > 970 and Heart_Rate > 68")
```
