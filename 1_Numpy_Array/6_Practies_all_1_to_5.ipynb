```python
import numpy as np
import time

# Numpy VS. Python Lists perfomance Test

# Python list
size = 1_000_000
list1 = list(range(size))
list2 = list(range(size))
start = time.time()
result = [x + y for x,y in zip(list1,list2)]
end = time.time()
print("Python list addition time:",end-start)

# Numpy Array
arr1 = np.array(list1)
arr2 = np.array(list2)
start = time.time()
result = arr1 + arr2
end = time.time()
print("Numpy array addition time:",end-start)
```

```python
# Memory Efficieny - Numpy VS Python List

import sys

list_data = list(range(1000))
Numpy_data = np.array(list_data)

print("Python list size:",sys.getsizeof(list_data)*len(list_data),"bytes")
print("Numpy araay size:",Numpy_data.nbytes,"bytes")
```

```python
# Indexing & Slicing in Numpy Array.

arr = np.array([10,20,30,40])  

print(arr[0])  #Indexing
print(arr[1])
print(arr[2])
print(arr[3])

# Slicing 

print(arr[0:5]) 
print(arr[0:2])
print(arr[3:4])
print(arr[:3])

# Indexing Negative

print(arr[-2])
print(arr[-4])
print(arr[-1])

# Slicing Negative

print(arr[-2:-1])
print(arr[:-1])
```

```python
# Fancy Indexing & Bo0lean Masking.

arr = np.array([50,30,25,65,78,39])
idx1 = [0,4,2]
idx2 = [1,3,5]
print("Print First index:",arr[idx1])
print("Print second index",arr[idx2])

# Boolen masking Condition valus grater than 33 pass. [Filer tha data]

marks = arr > 33
print("Print pass student grater than 33:",arr[marks])

# Boolen masking Condition valus less than 33 fail.
marks = arr < 33
print("Print fail student less than 33:",arr[marks])
```
