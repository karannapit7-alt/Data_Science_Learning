```python
import numpy as np
```

```python
# changing Data types:-

arr = np.array([1.,2.,3.,4.,5.,6.])
print("This is a float values:",arr)
print(arr.dtype)
```

```python
arr_int = arr.astype("int32")
print("This is integer values:",arr_int)
print(arr_int.dtype)
```

```python
# Downcasting to save memory.

arr_large = np.array([100000,200000,300000], dtype = np.int64)
print(arr_large)
print(arr_large.dtype)
arr_small = arr_large.astype(np.int32)
print(arr_small)
print(arr_small.dtype)
```

```python
# Memory usage:- 
arr_int64 = np.array([1,2,3], dtype=np.int64)
arr_int32 = np.array([1,2,3], dtype=np.int32)

print("bytes",arr_int64.nbytes)  # one elements 8 bytes.
print("bytes",arr_int32.nbytes)  # one elements 4 bytes.
```

```python
# Complex Number. 

arr = np.array([3 + 2j, 4 + 3j, 5 + 4j], dtype="complex128")
print(arr)
print(arr.dtype)

# Two sum complex Number.
arr1 = np.array([3 + 2j])
arr2 = np.array([4 + 3j])
print("Two Sum of Complex Number:",arr1 + arr2)

# Three sum of complex Number.
arr1 = np.array([3 + 2j])
arr2 = np.array([4 + 3j])
arr3 = np.array([5 + 4j])
print("Three Sum of Complex Number:",arr1 + arr2 + arr3)
```
