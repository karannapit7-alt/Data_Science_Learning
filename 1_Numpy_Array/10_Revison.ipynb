```python
import numpy as np
import time 
```

```python
# Adding Two Lists VS. Numpy Arrays.

# python List.
size = 1_000_000
list1 = list(range(size))
list2 = list(range(size))

start = time.time()
result = [x+y for x,y in zip(list1,list2)]
end = time.time()
print("Python list Time:",end-start)

# Numpy Arrays.

arr1 = np.array(list1)
arr2 = np.array(list2)

start = time.time()
result = arr1 + arr2
end = time.time()
print("Numpy Array Time:",end-start)
```

```python
# Memory efficieny- Numpys VS. Lists.
import sys 

list_data = list(range(11000))
numpy_data = np.array(list_data)

print("Python list size:",sys.getsizeof(list_data)*len(list_data),"bytes")
print("Nmupy Array size:",numpy_data.nbytes,"bytes")
```

```python
# Changing Data Types.

arr = np.array([1.2,5.3,4.5])
print(arr)
print(arr.dtype)

int_arr = arr.astype("int64")
print(int_arr)
print(int_arr.dtype)
```

```python
# Downcasting to save memory.

arr_large = np.array([10000,20000,30000])
print(arr_large)
print(arr_large.dtype)

arr_small = arr_large.astype(np.int32)
print(arr_small)
print(arr_small.dtype)
```

```python
# memory usage:- 

arr_int64 = np.array([10,20,30], dtype=np.int64)
arr_int32 = np.array([10,20,30], dtype=np.int32)

print(arr_int64)
print(arr_int64.nbytes)

print(arr_int32)
print(arr_int32.nbytes)
```

```python
# Complex Number in Numpy Arrays.

complex_arr = np.array([1+3j, 2+4j, 3+10j])
print(complex_arr)

# Sum Two number.

arr1 = np.array([1+3j])
arr2 = np.array([2+4j])
print("Sum Two number",arr1 + arr2)

# Sum Three number.

arr1 = np.array([1+3j])
arr2 = np.array([2+4j])
arr3 = np.array([3+10j])
print("Sum Three number",arr1 + arr2 + arr3)
```

```python
# BoardCasting in Numpy Array.

square = np.array([1,2,3,4,5])
print(square)
result = []

for num in square:
    result.append(num**2)
print("Square",result)
```

```python
# Vectorized opertion.

arr = np.array([1,2,3,4,5])
square = arr**2
print(square)
```

```python
# BroadCasting: Scling Arrays without extra memory.

arr = np.array([1,2,3,4,5])
print(arr)
result = arr + 10
print(result)
```

```python
# BoardCasting with Two Arrays.

arr1 = np.array([1,2,3,4])
arr2 = np.array([7,1,2,4])

arr_sum = arr1 + arr2
print(arr_sum)
```

```python
# Boardcasting a 2D Array and 1D Array.

arr1 = np.array([[1,2,3],[4,5,6]])
arr2 = np.array([10,20,30])

result = arr1 + arr2
print(result)
```
