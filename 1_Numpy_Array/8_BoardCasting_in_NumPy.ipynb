```python
import numpy as np
```

```python
arr = np.array([1,2,3,4,5,6])
print(arr)
result = []

for num in arr:
    result.append(num**2)
print(result)
```

```python
# Vectorized Operation.

arr = np.array([1,2,3,4,5,6])
result = arr ** 2
print(result)
```

```python
# Broadcasting Scaling Arrays without Extra Memory.

arr = np.array([10,20,30,40,50,60])
result = arr + 12
print(result)
```

```python
# Boardcasting with Two Arrays.

# logic step
arr1 = np.array([10,20,30])
arr2 = np.array([2,4,7])
print(arr1 + arr2)

# more logic step 
arr3 = np.array([2,7,5,3])
arr4 = np.array([5,2,3,8])
result = arr3 * arr4
print(result)
```

```python
arr = np.array([[1,2,3],[4,5,6]])
print(arr)
arr1 = arr + ([1,2,3])
print("2D Array sum:",arr1)
```

```python
arr1 = np.array([[1,2,3],[4,5,6]])
arr2 = np.array([10,20,30])
result = arr1 + arr2
print(result)
```

```python
# Normalizing Data using Broadcasting.

Data = np.array([[10,20,30],[15,25,35],[20,35,45],[30,40,50]])
print(Data)

mean = Data.mean(axis=0)
std = Data.std(axis=0)

normalized_Data = (Data - mean) / std
print(normalized_Data)
```
